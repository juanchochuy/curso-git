```python
from datetime import datetime
from dateutil.relativedelta import relativedelta
import pandas as pd

# Datos proporcionados
data = [
    ["06/07/2025", 372.33, 744.68, 2, "octubre 2025"],
    ["12/06/2025", 739.15, 4434.92, 6, "febrero 2026"],
    ["01/04/2025", 117.50, 1997.50, 17, "enero 2027"],
    ["30/03/2025", 1018.33, 11201.69, 11, "julio 2026"],
    ["02/03/2025", 127.75, 3577.00, 28, "diciembre 2027"],
    ["31/12/2024", 289.90, 289.90, 1, "octubre 2025"],
    ["22/12/2024", 277.72, 7220.80, 26, "octubre 2027"],
    ["13/10/2024", 199.83, 1199.04, 6, "febrero 2026"],
    ["03/12/2023", 44.42, 621.76, 14, "octubre 2026"],
    ["06/08/2023", 380.47, 3804.78, 10, "junio 2026"],
    ["17/07/2023", 94.39, 849.47, 9, "mayo 2026"],
]

# Crear DataFrame
df = pd.DataFrame(data, columns=["Fecha Compra", "Cuota", "Saldo", "Meses Restantes", "Fecha Final Estimada"])

# Mes de inicio: septiembre 2025
start_date = datetime(2025, 9, 1)

# Crear un diccionario para acumular las cuotas que dejarán de pagarse por mes
monthly_reduction = {}

# Para cada fila, sumar la cuota al mes siguiente de su término
for _, row in df.iterrows():
    cuota = row["Cuota"]
    meses_restantes = int(row["Meses Restantes"])
    end_month = start_date + relativedelta(months=meses_restantes)
    month_str = end_month.strftime("%Y-%m")
    if month_str in monthly_reduction:
        monthly_reduction[month_str] += cuota
    else:
        monthly_reduction[month_str] = cuota

# Convertir a DataFrame y ordenar por fecha
reduction_df = pd.DataFrame(monthly_reduction.items(), columns=["Mes", "Reducción de Pago"])
reduction_df["Mes"] = pd.to_datetime(reduction_df["Mes"])
reduction_df = reduction_df.sort_values("Mes")
reduction_df["Mes"] = reduction_df["Mes"].dt.strftime("%B %Y")

# Mostrar resultado
print(reduction_df.reset_index(drop=True))
```
